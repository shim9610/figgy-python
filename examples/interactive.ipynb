## Figgy interactive plot
Install `figgy-python[notebook]`, run the cell, select objects, edit properties and copy the resulting Python settings.

```python
import numpy as np
import figgy

x = np.linspace(0, 20, 100_000)
fig, ax = figgy.subplots(backend="webgpu")
ax.plot(x, np.sin(x), id="signal", label="signal")
ax.set_title("Edit this title in the notebook")
ax.set_xlabel("Time (s)")
ax.legend()
display(fig.widget())
```

```python
await fig.sync()
print(fig.to_code())
await fig.savefig_async("edited.png")
```
